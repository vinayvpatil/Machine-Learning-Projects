    **ax=ax:** Explicitly tells Seaborn to draw the plot on the current subplot Axes object.  
    **saturation=0.95:** Adjusts the color intensity.   
    **for container in ax.containers: ...:** This inner loop iterates over the bar containers in each plot.   
    **ax.bar_label(container, ...):** This Matplotlib function adds text labels with the count values on top of each bar, making the plot easier to read.  
    **ax.set_title(...), ax.set_xlabel(...), ax.set_ylabel(...):** These methods set the title and axis labels for the current subplot.

```python
# checking columns
df_cleaned.columns
```

```python
# features 'year', 'mileage', 'tax', 'mpg', 'engineSize' needs to be scaled (standardized)
cols = ['year', 'mileage', 'tax', 'mpg', 'engineSize']

scaler = StandardScaler()
df_cleaned[cols] = StandardScaler().fit_transform(df_cleaned[cols])
```

**StandardScaler** is a class from scikit-learn's preprocessing module. This class is a tool for standardizing features.  
**cols = ['age','bmi','children']:** This creates a Python list of strings containing the names of the columns we want to standardize.  
**df_cleaned[cols]:** This selects only the columns from the df_cleaned DataFrame that are specified in the cols list.   
**StandardScaler():** This creates a new instance of the StandardScaler class. It's like preparing a tool to do a specific job.  
**.fit_transform():** This is a powerful method that combines two steps into one:  
**.fit():** This scaler calculates the mean and standard deviation for each of the selected columns ('age', 'bmi', 'children').   
**.transform():** Using the mean and standard deviation it just calculated, this scaler transforms every value in those columns using the following formula:   
z=(x- μ) / σ    
where: z is the standardized value, x is the original value, μ is the mean of the feature column, σ is the standard deviation of the feature  
    
**column.df_cleaned[cols]** = : The final result is a new NumPy array containing the standardized values.    
This transformed array is then used to overwrite the original 'age', 'bmi', and 'children' columns in the df_cleaned DataFrame.  

**Why this is done in machine learning**    
Standardization is a common data preprocessing step, and it's essential for several reasons:    
Equal importance for all features: Many machine learning algorithms are sensitive to the scale and range of input features. If one feature has a much larger range of values than another (e.g., age in decades vs. bmi in the tens), the larger-scale feature could disproportionately influence the model.  Standardization ensures that all features are on a comparable scale.     
Faster convergence: For algorithms that use gradient descent (like neural networks and logistic regression), standardization can lead to faster and more stable convergence.         
Improved performance: Many algorithms, particularly those based on distance calculations like K-Nearest Neighbors (KNN), assume that all features are centered around zero and have similar variance. Standardizing the data helps these models perform better.

# <div style="color:blue;display:inline-block;border-radius:5px;background-color:#F0E68C;font-family:Nexa;overflow:hidden"><p style="padding:15px;color:blue;overflow:hidden;font-size:100%;letter-spacing:0.5px;margin:0; width:750px;"><b> </b> 6) Splitting the Data </p></div>

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

This function splits the data into four subsets:  
X_train: 80% of the independent variables for training the model.  
X_test: 20% of the independent variables for testing the model.  
y_train: 80% of the dependent variable for training.  
y_test: 20% of the dependent variable for testing.  
The **test_size=0.20** argument specifies that 20% of the data should be reserved for the test set.   
The **random_state=42** argument ensures that the split is reproducible, meaning you will get the exact same split of data every time you run the code. 

# <div style="color:blue;display:inline-block;border-radius:5px;background-color:#F0E68C;font-family:Nexa;overflow:hidden"><p style="padding:15px;color:blue;overflow:hidden;font-size:100%;letter-spacing:0.5px;margin:0; width:750px;"><b> </b> 7) Model Training </p></div>

```python
Linear_reg = LinearRegression()
Ride_reg= Ridge()
XGBoost = XGBRegressor()
Random_forest = RandomForestRegressor()
Ada_boost = AdaBoostRegressor()
Grad_boost = GradientBoostingRegressor()
Bagging= BaggingRegressor()
Decision_tree = DecisionTreeRegressor()
Svr = SVR()
Knn = KNeighborsRegressor()
```

```python
model_li = [Ride_reg,XGBoost,Random_forest,Ada_boost,Grad_boost,Bagging, Decision_tree,Svr,Knn]

Score =[]
for model in model_li:
    model.fit(X_train,y_train)
    Score.append(r2_score(y_test,model.predict(X_test)))

Score
# The model is trained (or "fitted") on the training data. The model learns the relationship between 
# the features (X_train) and the target variable (y_train) to find the best-fit line.
```

# <div style="color:blue;display:inline-block;border-radius:5px;background-color:#F0E68C;font-family:Nexa;overflow:hidden"><p style="padding:15px;color:blue;overflow:hidden;font-size:100%;letter-spacing:0.5px;margin:0; width:750px;"><b> </b> 8) Model Evaluation </p></div>

R squared value of test data

```python
model_names = ["Linear Regression", "Ridge Regression", "XGBoost", "Random Forest", "AdaBoost", "Gradient Boosting", "Bagging", "Decision Tree", "SVR", "KNN"]

plt.figure(figsize=(15, 10))

# Convert X_test back to DataFrame for plotting
X_test_df = pd.DataFrame(X_test, columns=X.columns)

for i, model in enumerate(model_li):
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    r2 = r2_score(y_test, y_pred)
    Score.append(r2)

    plt.subplot(3, 3, i+1)
    plt.scatter(X_test_df['engineSize'], y_test, color='blue', label="Actual")
    plt.scatter(X_test_df['engineSize'], y_pred, color='red', label="Predicted", alpha=0.7)
    plt.plot(np.sort(X_test_df['engineSize']), np.sort(y_pred), color='green', linewidth=2)
    plt.title(f"{model_names[i]} (R²={r2:.2f})")
    plt.xlabel("Engine Size")
    plt.ylabel("Price")
    plt.legend()

plt.tight_layout()
plt.show()

for name, score in zip(model_names, Score):
    print(f"{name}: R² Score = {score:.4f}")
```

**Interpretation of test metrics**  
**Test R-squared**: This value tells you how much of the variance in your test data is explained by your model. It is the most direct measure of your model's predictive power on unseen data.
